```python
# -*- coding: utf-8 -*-
"""
4DGS-JEPA — Experiment 4: Held-Out Next-Frame Rendering (Route B)

Controlled rendered-video bridge:
  Gaussian trajectory -> causal JEPA prediction -> future Gaussian geometry -> rendered frame.

The JEPA is NOT trained with RGB/rendering loss. RGB and opacity are fixed through
time and transported from the source Gaussian state. The final frame of every
trajectory is excluded from all training/validation targets. Test trajectories are
disjoint from train/validation trajectories.

Main test:
    G_{T-L:T-1} -> Z_{T-1} -> Zhat_T -> Ghat_T -> Ihat_T
and compare Ihat_T with the held-out I_T.

Training remains multi-horizon with {1,2,4}, so temporal composition is non-trivial.
A secondary held-out Delta=4 diagnostic compares one direct transition with a
composed (2,2) rollout to the same unseen terminal frame.
"""

import math, json, random, shutil, copy
from pathlib import Path
from dataclasses import dataclass, asdict
from typing import Dict, List, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

try:
    from IPython.display import display
except Exception:
    display = print


# ============================================================
# Setup
# ============================================================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
try:
    torch.set_float32_matmul_precision("high")
except Exception:
    pass


def seed_all(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


@dataclass
class Config:
    # Small synthetic rendered-video dataset
    num_groups: int = 3
    gaussians_per_group: int = 6
    seq_len: int = 24
    context_len: int = 4
    train_scenes: int = 128
    val_scenes: int = 16
    test_scenes: int = 20

    # Model
    token_dim: int = 48
    horizon_dim: int = 16
    transformer_heads: int = 4
    transformer_layers: int = 1
    transformer_dropout: float = 0.0

    # Optimization
    batch_size: int = 48
    eval_batch_size: int = 64
    epochs: int = 15
    lr: float = 1e-3
    min_lr: float = 1e-5
    weight_decay: float = 1e-5
    ema_tau: float = 0.99
    grad_clip: float = 1.0

    lambda_geo: float = 5.0
    lambda_path: float = 1.0
    lambda_comp: float = 0.5
    lambda_rep: float = 0.02
    lambda_cov_rep: float = 0.05
    var_floor: float = 0.5
    center_huber_delta: float = 0.10
    cov_loss_weight: float = 0.25

    # Fixed causal-feature scales inherited from Exp1/Exp2
    position_scale: float = 1.5
    relative_position_scale: float = 0.20
    velocity_scale: float = 0.05
    acceleration_scale: float = 0.01
    cov_delta_scale: float = 0.05

    # Multi-horizon training; the held-out final test remains Delta=1
    train_horizons: Tuple[int, ...] = (1, 2, 4)

    # Renderer
    image_size: int = 96
    focal_px: float = 95.0
    camera_distance: float = 4.0
    min_projected_var: float = 0.35
    alpha_cap: float = 0.995
    background: float = 0.03

    # Repetition / qualitative output
    seeds: Tuple[int, ...] = (0, 1, 2)
    highlight_variant: str = "Endpoint + Composition"
    highlight_seed: int = 0
    highlight_scene_index: int = 0

    # Optional; requires `pip install lpips` and pretrained weights.
    compute_lpips: bool = False


CFG = Config()
seed_all(0)
print(CFG)

BASE_DIR = Path("/content") if Path("/content").exists() else Path.cwd()
RESULT_DIR = BASE_DIR / "4dgs_jepa_exp4_video_results"
if RESULT_DIR.exists():
    shutil.rmtree(RESULT_DIR)
RESULT_DIR.mkdir(parents=True, exist_ok=True)
with open(RESULT_DIR / "config.json", "w") as f:
    json.dump(asdict(CFG), f, indent=2)


# ============================================================
# Geometry utilities
# ============================================================

def skew(v):
    x, y, z = v.unbind(-1)
    O = torch.zeros_like(x)
    return torch.stack([O, -z, y, z, O, -x, -y, x, O], dim=-1).reshape(*v.shape[:-1], 3, 3)


def axis_angle_to_matrix(w):
    W = skew(w)
    theta2 = (w * w).sum(dim=-1, keepdim=True)
    theta = torch.sqrt(theta2.clamp_min(1e-16))
    small = theta2 < 1e-6
    A = torch.where(
        small,
        1.0 - theta2 / 6.0 + theta2 * theta2 / 120.0,
        torch.sin(theta) / theta.clamp_min(1e-8),
    )
    B = torch.where(
        small,
        0.5 - theta2 / 24.0 + theta2 * theta2 / 720.0,
        (1.0 - torch.cos(theta)) / theta2.clamp_min(1e-8),
    )
    I = torch.eye(3, device=w.device, dtype=w.dtype).expand(*w.shape[:-1], 3, 3)
    return I + A[..., None] * W + B[..., None] * (W @ W)


def np_axis_angle_to_matrix(w):
    theta = float(np.linalg.norm(w))
    if theta < 1e-10:
        K = np.array([[0, -w[2], w[1]], [w[2], 0, -w[0]], [-w[1], w[0], 0]], np.float32)
        return np.eye(3, dtype=np.float32) + K
    a = w / theta
    K = np.array([[0, -a[2], a[1]], [a[2], 0, -a[0]], [-a[1], a[0], 0]], np.float32)
    return (np.eye(3, dtype=np.float32) + np.sin(theta) * K + (1 - np.cos(theta)) * (K @ K)).astype(np.float32)


def make_spd(log_scales):
    return np.diag(np.exp(2.0 * log_scales).astype(np.float32))


def cov_to_log_cholesky(cov):
    I = torch.eye(3, dtype=cov.dtype, device=cov.device)
    L = torch.linalg.cholesky(cov + 1e-6 * I)
    diag = torch.diagonal(L, dim1=-2, dim2=-1).clamp_min(1e-8).log()
    off = torch.stack([L[..., 1, 0], L[..., 2, 0], L[..., 2, 1]], dim=-1)
    return torch.cat([diag, off], dim=-1)


# ============================================================
# Synthetic dynamic Gaussian videos
# ============================================================

def generate_scene(seed: int, cfg: Config) -> Dict[str, np.ndarray]:
    rng = np.random.default_rng(seed)
    G, P, T = cfg.num_groups, cfg.gaussians_per_group, cfg.seq_len
    N = G * P
    group_ids = np.repeat(np.arange(G), P).astype(np.int64)

    group_centers = rng.uniform(-1.0, 1.0, (G, 3)).astype(np.float32)
    group_centers[:, 2] *= 0.55

    base_mu = np.zeros((N, 3), np.float32)
    base_cov = np.zeros((N, 3, 3), np.float32)
    local_phase = rng.uniform(0, 2*np.pi, (N, 1)).astype(np.float32)
    local_dir = rng.normal(size=(N, 3)).astype(np.float32)
    local_dir /= np.linalg.norm(local_dir, axis=1, keepdims=True) + 1e-8
    local_amp = rng.uniform(0.0, 0.035, (N, 1)).astype(np.float32)
    local_freq = rng.uniform(0.10, 0.35, (N, 1)).astype(np.float32)
    scale_phase = rng.uniform(0, 2*np.pi, (N, 3)).astype(np.float32)
    scale_amp = rng.uniform(0.0, 0.06, (N, 3)).astype(np.float32)

    for i in range(N):
        g = group_ids[i]
        base_mu[i] = group_centers[g] + rng.normal(0, 0.16, 3)
        base_mu[i, 2] = np.clip(base_mu[i, 2], -0.8, 0.8)
        base_cov[i] = make_spd(rng.uniform(np.log(0.045), np.log(0.11), 3))

    axes = rng.normal(size=(G, 3)).astype(np.float32)
    axes /= np.linalg.norm(axes, axis=1, keepdims=True) + 1e-8
    omega = rng.uniform(-0.045, 0.045, (G, 1)).astype(np.float32)
    ang_acc = rng.uniform(-0.002, 0.002, (G, 1)).astype(np.float32)
    velocity = rng.uniform(-0.035, 0.035, (G, 3)).astype(np.float32)
    accel = rng.uniform(-0.0015, 0.0015, (G, 3)).astype(np.float32)
    trans_amp = rng.uniform(0.0, 0.08, (G, 3)).astype(np.float32)
    trans_freq = rng.uniform(0.08, 0.18, (G, 1)).astype(np.float32)
    trans_phase = rng.uniform(0, 2*np.pi, (G, 3)).astype(np.float32)

    # Smaller depth changes keep the rendered objects stably visible.
    velocity[:, 2] *= 0.35
    accel[:, 2] *= 0.35
    trans_amp[:, 2] *= 0.35

    if seed % 2 == 0:
        omega[0] = ang_acc[0] = 0
        velocity[0] = accel[0] = trans_amp[0] = 0

    mu = np.zeros((T, N, 3), np.float32)
    cov = np.zeros((T, N, 3, 3), np.float32)

    for t in range(T):
        tt = float(t)
        Rg_all, dg_all = [], []
        for g in range(G):
            angle = omega[g, 0] * tt + 0.5 * ang_acc[g, 0] * tt * tt
            Rg = np_axis_angle_to_matrix(axes[g] * angle)
            dg = velocity[g] * tt + 0.5 * accel[g] * tt * tt + trans_amp[g] * np.sin(trans_freq[g, 0] * tt + trans_phase[g])
            Rg_all.append(Rg)
            dg_all.append(dg)

        for i in range(N):
            g = group_ids[i]
            r = local_amp[i] * np.sin(local_freq[i] * tt + local_phase[i]) * local_dir[i]
            log_s = scale_amp[i] * np.sin(0.12 * tt + scale_phase[i])
            U = np.diag(np.exp(log_s).astype(np.float32))
            Rg, dg = Rg_all[g], dg_all[g]
            mu[t, i] = Rg @ base_mu[i] + dg + r
            cov[t, i] = Rg @ U @ base_cov[i] @ U.T @ Rg.T

    # Fixed appearance: rendering evaluates predicted geometry, not appearance learning.
    palette = np.array(
        [[0.88, 0.28, 0.22], [0.20, 0.68, 0.92], [0.32, 0.82, 0.42], [0.86, 0.62, 0.18]],
        np.float32,
    )
    rgb = np.zeros((N, 3), np.float32)
    for i in range(N):
        rgb[i] = np.clip(palette[group_ids[i] % len(palette)] + rng.normal(0, 0.06, 3), 0.05, 0.98)
    opacity = rng.uniform(0.55, 0.82, N).astype(np.float32)

    return {"mu": mu, "cov": cov, "group_ids": group_ids, "rgb": rgb, "opacity": opacity}


class GaussianVideoTrainDataset(Dataset):
    """Final frame is never used as a training/validation target."""
    def __init__(self, scene_seeds: List[int], cfg: Config):
        self.cfg = cfg
        self.scene_seeds = list(scene_seeds)
        self.scenes = [generate_scene(s, cfg) for s in scene_seeds]
        self.heldout_idx = cfg.seq_len - 1
        self.max_h = max(cfg.train_horizons)
        self.samples = []
        # Last permitted t satisfies t + max_h <= heldout_idx - 1.
        for si in range(len(self.scenes)):
            for t in range(cfg.context_len - 1, self.heldout_idx - self.max_h):
                self.samples.append((si, t))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        si, t = self.samples[idx]
        sc = self.scenes[si]
        return {
            "mu": torch.from_numpy(sc["mu"]),
            "cov": torch.from_numpy(sc["cov"]),
            "group_ids": torch.from_numpy(sc["group_ids"]),
            "t": torch.tensor(t, dtype=torch.long),
        }


def build_splits(cfg):
    return (
        list(range(10_000, 10_000 + cfg.train_scenes)),
        list(range(20_000, 20_000 + cfg.val_scenes)),
        list(range(30_000, 30_000 + cfg.test_scenes)),
    )


# ============================================================
# Evaluation-only Gaussian renderer
# ============================================================

@torch.no_grad()
def render_gaussians(mu, cov, rgb, opacity, cfg):
    """Perspective projection + projected covariance + alpha compositing."""
    dtype, device = mu.dtype, mu.device
    H = W = cfg.image_size
    f = float(cfg.focal_px)
    cx = (W - 1) / 2.0
    cy = (H - 1) / 2.0

    # Camera at (0,0,+d), optical axis toward -world-z.
    Rcw = torch.tensor([[1.,0.,0.],[0.,-1.,0.],[0.,0.,-1.]], dtype=dtype, device=device)
    cam_pos = torch.tensor([0.,0.,float(cfg.camera_distance)], dtype=dtype, device=device)
    xyz = (Rcw @ (mu - cam_pos).T).T
    cov_cam = Rcw[None] @ cov @ Rcw.T[None]

    x, y, z = xyz.unbind(-1)
    valid = z > 0.15
    z = z.clamp_min(0.15)
    u = f * x / z + cx
    v = f * y / z + cy

    J = torch.zeros(mu.shape[0], 2, 3, dtype=dtype, device=device)
    J[:, 0, 0] = f / z
    J[:, 0, 2] = -f * x / z.square()
    J[:, 1, 1] = f / z
    J[:, 1, 2] = -f * y / z.square()
    Sigma2 = J @ cov_cam @ J.transpose(-1, -2)
    Sigma2 = Sigma2 + float(cfg.min_projected_var) * torch.eye(2, dtype=dtype, device=device)[None]
    inv2 = torch.linalg.inv(Sigma2)

    yy, xx = torch.meshgrid(
        torch.arange(H, dtype=dtype, device=device),
        torch.arange(W, dtype=dtype, device=device),
        indexing="ij",
    )
    accum = torch.zeros((H, W, 3), dtype=dtype, device=device)
    trans = torch.ones(H, W, dtype=dtype, device=device)

    for idx in torch.argsort(z).tolist():  # front-to-back: smaller camera-z is closer
        if not bool(valid[idx]):
            continue
        delta = torch.stack([xx - u[idx], yy - v[idx]], dim=-1)
        q = torch.einsum("...i,ij,...j->...", delta, inv2[idx], delta)
        a = opacity[idx] * torch.exp(-0.5 * q)
        a = torch.where(q <= 18.0, a, torch.zeros_like(a)).clamp(0, cfg.alpha_cap)
        contrib = trans * a
        accum = accum + contrib[..., None] * rgb[idx]
        trans = trans * (1.0 - a)

    image = accum + trans[..., None] * float(cfg.background)
    return image.permute(2, 0, 1).clamp(0, 1)


def render_scene_frame(scene, t, cfg):
    return render_gaussians(
        torch.from_numpy(scene["mu"][t]).to(DEVICE),
        torch.from_numpy(scene["cov"][t]).to(DEVICE),
        torch.from_numpy(scene["rgb"]).to(DEVICE),
        torch.from_numpy(scene["opacity"]).to(DEVICE),
        cfg,
    )


# ============================================================
# Image metrics
# ============================================================

def image_mse(x, y):
    return float(F.mse_loss(x, y).cpu())


def image_psnr(x, y):
    mse = F.mse_loss(x, y).clamp_min(1e-12)
    return float((-10 * torch.log10(mse)).cpu())


def image_ssim(x, y, win=7):
    x, y = x[None], y[None]
    pad = win // 2
    mux = F.avg_pool2d(x, win, 1, pad)
    muy = F.avg_pool2d(y, win, 1, pad)
    vx = (F.avg_pool2d(x*x, win, 1, pad) - mux*mux).clamp_min(0)
    vy = (F.avg_pool2d(y*y, win, 1, pad) - muy*muy).clamp_min(0)
    vxy = F.avg_pool2d(x*y, win, 1, pad) - mux*muy
    c1, c2 = 0.01**2, 0.03**2
    s = ((2*mux*muy+c1)*(2*vxy+c2))/((mux.square()+muy.square()+c1)*(vx+vy+c2))
    return float(s.mean().cpu())


def maybe_build_lpips(cfg):
    if not cfg.compute_lpips:
        return None
    try:
        import lpips
        return lpips.LPIPS(net="alex").to(DEVICE).eval()
    except Exception as exc:
        print("LPIPS unavailable; continuing without it:", exc)
        return None


@torch.no_grad()
def lpips_score(net, x, y):
    if net is None:
        return float("nan")
    return float(net((2*x-1)[None], (2*y-1)[None]).mean().cpu())


# ============================================================
# Causal encoder / JEPA / selective geometry decoder
# ============================================================

def _normalize_cov_coord(coord):
    return torch.cat([(coord[..., :3] - math.log(0.07))/0.5, coord[..., 3:]/0.05], dim=-1)


def causal_gaussian_features(mu, cov, group_ids, cfg):
    B, L, N, _ = mu.shape
    assert L == cfg.context_len and L >= 3
    cov_coord = cov_to_log_cholesky(cov)
    cur_mu, cur_cov = mu[:, -1], cov_coord[:, -1]

    group_centers = []
    for g in range(cfg.num_groups):
        mask = (group_ids == g).float().unsqueeze(-1)
        group_centers.append((cur_mu * mask).sum(1) / mask.sum(1).clamp_min(1.0))
    group_centers = torch.stack(group_centers, dim=1)
    bidx = torch.arange(B, device=mu.device)[:, None]
    local_pos = cur_mu - group_centers[bidx, group_ids]

    geom = torch.cat(
        [
            cur_mu/cfg.position_scale,
            local_pos/cfg.relative_position_scale,
            _normalize_cov_coord(cur_cov),
            torch.ones(B, N, 1, device=mu.device, dtype=mu.dtype),
            F.one_hot(group_ids, num_classes=cfg.num_groups).float(),
        ],
        dim=-1,
    )

    rel_hist = ((mu[:, :-1] - cur_mu[:, None]) / cfg.relative_position_scale).permute(0,2,1,3).reshape(B,N,-1)
    velocity = (mu[:, -1] - mu[:, -2]) / cfg.velocity_scale
    prev_velocity = (mu[:, -2] - mu[:, -3]) / cfg.velocity_scale
    acceleration = (velocity - prev_velocity) * (cfg.velocity_scale/cfg.acceleration_scale)
    delta_cov = (cov_coord[:, -1] - cov_coord[:, -2]) / cfg.cov_delta_scale
    motion = torch.cat([rel_hist, velocity, acceleration, delta_cov], dim=-1)
    return geom, motion


class HierEncoder(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D = cfg.token_dim
        geom_dim = 3 + 3 + 6 + 1 + cfg.num_groups
        motion_dim = 3*(cfg.context_len-1) + 3 + 3 + 6
        self.cfg = cfg
        self.geom_mlp = nn.Sequential(nn.Linear(geom_dim,D), nn.GELU(), nn.Linear(D,D))
        self.motion_mlp = nn.Sequential(nn.Linear(motion_dim,D), nn.GELU(), nn.Linear(D,D))
        self.fuse = nn.Sequential(nn.LayerNorm(D), nn.Linear(D,D), nn.GELU())
        layer = nn.TransformerEncoderLayer(
            D, cfg.transformer_heads, dim_feedforward=2*D, batch_first=True,
            norm_first=True, dropout=cfg.transformer_dropout
        )
        self.contextualizer = nn.TransformerEncoder(layer, cfg.transformer_layers)
        self.type_embed = nn.Parameter(torch.randn(3,D)*0.02)
        self.norm = nn.LayerNorm(D)

    def forward(self, mu_hist, cov_hist, group_ids):
        geom, motion = causal_gaussian_features(mu_hist, cov_hist, group_ids, self.cfg)
        gtok = self.fuse(self.geom_mlp(geom) + self.motion_mlp(motion))
        groups = []
        for g in range(self.cfg.num_groups):
            mask = (group_ids == g).float().unsqueeze(-1)
            groups.append((gtok*mask).sum(1)/mask.sum(1).clamp_min(1.0))
        otok = torch.stack(groups, dim=1)
        stok = otok.mean(dim=1, keepdim=True)
        tokens = torch.cat(
            [stok+self.type_embed[0], otok+self.type_embed[1], gtok+self.type_embed[2]],
            dim=1,
        )
        return self.norm(self.contextualizer(tokens))


class HorizonEmbedding(nn.Module):
    def __init__(self, out_dim, cfg):
        super().__init__()
        self.scale = float(max(cfg.train_horizons))
        self.mlp = nn.Sequential(nn.Linear(4,out_dim), nn.GELU(), nn.Linear(out_dim,out_dim))

    def forward(self, h):
        h = h.float().view(-1,1)
        return self.mlp(torch.cat([h/self.scale, torch.sin(h), torch.cos(h), torch.log1p(h)], dim=-1))


class Transition(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D = cfg.token_dim
        self.h = HorizonEmbedding(D, cfg)
        layer = nn.TransformerEncoderLayer(
            D, cfg.transformer_heads, dim_feedforward=2*D, batch_first=True,
            norm_first=True, dropout=cfg.transformer_dropout
        )
        self.net = nn.TransformerEncoder(layer, cfg.transformer_layers)
        self.out = nn.Sequential(nn.LayerNorm(D), nn.Linear(D,D))

    def forward(self, z, h):
        return z + self.out(self.net(z + self.h(h)[:,None,:]))


class GeometryDecoder(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D, H = cfg.token_dim, cfg.horizon_dim
        self.cfg = cfg
        self.h = HorizonEmbedding(H, cfg)
        self.group_head = nn.Sequential(nn.Linear(2*D+H,2*D), nn.GELU(), nn.Linear(2*D,6))
        self.gauss_head = nn.Sequential(nn.Linear(2*D+H,2*D), nn.GELU(), nn.Linear(2*D,6))

    def forward(self, z0, z1, source_mu, source_cov, group_ids, h):
        B, N, _ = source_mu.shape
        G = self.cfg.num_groups
        he = self.h(h)
        go = self.group_head(torch.cat([z0[:,1:1+G], z1[:,1:1+G], he[:,None].expand(B,G,-1)], dim=-1))
        hscale = h.float().view(B,1,1)
        aa = (0.10*hscale)*torch.tanh(go[...,:3])
        d = (0.10*hscale)*torch.tanh(go[...,3:])
        R = axis_angle_to_matrix(aa)

        g0, g1 = z0[:,1+G:1+G+N], z1[:,1+G:1+G+N]
        po = self.gauss_head(torch.cat([g0,g1,he[:,None].expand(B,N,-1)], dim=-1))
        r = (0.025*hscale)*torch.tanh(po[...,:3])
        log_s = (0.04*hscale)*torch.tanh(po[...,3:])

        bidx = torch.arange(B, device=source_mu.device)[:,None]
        Ri, di = R[bidx,group_ids], d[bidx,group_ids]
        pred_mu = (Ri @ source_mu.unsqueeze(-1)).squeeze(-1) + di + r
        U = torch.diag_embed(torch.exp(log_s))
        pred_cov = Ri @ U @ source_cov @ U.transpose(-1,-2) @ Ri.transpose(-1,-2)
        return {"mu":pred_mu, "cov":pred_cov, "R":R, "d":d, "r":r, "log_s":log_s}


class JEPA(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.encoder = HierEncoder(cfg)
        self.transition = Transition(cfg)
        self.decoder = GeometryDecoder(cfg)


# ============================================================
# Losses / partitions / rollout
# ============================================================

def hier_distance(a,b,cfg):
    G = cfg.num_groups
    s = 1 - F.cosine_similarity(a[:,0],b[:,0],dim=-1).mean()
    o = (1 - F.cosine_similarity(a[:,1:1+G],b[:,1:1+G],dim=-1)).mean()
    g = F.smooth_l1_loss(a[:,1+G:],b[:,1+G:])
    return s+o+g


def hier_distance_per_sample(a,b,cfg):
    G = cfg.num_groups
    s = 1 - F.cosine_similarity(a[:,0],b[:,0],dim=-1)
    o = (1 - F.cosine_similarity(a[:,1:1+G],b[:,1:1+G],dim=-1)).mean(-1)
    g = F.smooth_l1_loss(a[:,1+G:],b[:,1+G:],reduction="none").mean((-1,-2))
    return s+o+g


def huber_center(pred,tgt,delta):
    dist = torch.linalg.norm(pred-tgt,dim=-1)
    d = torch.as_tensor(delta,dtype=dist.dtype,device=dist.device)
    return torch.where(dist<=d,0.5*dist.square(),d*(dist-0.5*d)).mean()


def geometry_loss(pred_mu,pred_cov,tgt_mu,tgt_cov,cfg):
    return huber_center(pred_mu,tgt_mu,cfg.center_huber_delta) + cfg.cov_loss_weight*F.smooth_l1_loss(
        cov_to_log_cholesky(pred_cov), cov_to_log_cholesky(tgt_cov)
    )


def covariance_error_per_sample(pred_cov,tgt_cov):
    return torch.linalg.norm(cov_to_log_cholesky(pred_cov)-cov_to_log_cholesky(tgt_cov),dim=-1).mean(-1)


def _vicreg_level(x,cfg):
    x = x.reshape(-1,x.shape[-1])
    std = torch.sqrt(x.var(0,unbiased=False)+1e-4)
    lvar = F.relu(cfg.var_floor-std).square().mean()
    xc = x-x.mean(0,keepdim=True)
    cov = (xc.T@xc)/max(1,x.shape[0]-1)
    off = cov-torch.diag(torch.diagonal(cov))
    return lvar + cfg.lambda_cov_rep*off.square().sum()/x.shape[1]


def rep_regularizer(z,cfg):
    G = cfg.num_groups
    return (_vicreg_level(z[:,:1],cfg)+_vicreg_level(z[:,1:1+G],cfg)+_vicreg_level(z[:,1+G:],cfg))/3


TRAIN_PARTITIONS = {1:[[1]], 2:[[1,1]], 4:[[2,2],[1,1,2]]}
PRIMARY_PARTITIONS = {1:[1], 2:[1,1], 4:[2,2]}


def get_hist(batch,h_end,cfg):
    mus,covs = [],[]
    for b,tend in enumerate(h_end.tolist()):
        start = tend-cfg.context_len+1
        mus.append(batch["mu"][b,start:tend+1])
        covs.append(batch["cov"][b,start:tend+1])
    return torch.stack(mus),torch.stack(covs)


def scene_hist(scene,end_idx,cfg):
    start = end_idx-cfg.context_len+1
    return (
        torch.from_numpy(scene["mu"][start:end_idx+1]).unsqueeze(0).to(DEVICE),
        torch.from_numpy(scene["cov"][start:end_idx+1]).unsqueeze(0).to(DEVICE),
        torch.from_numpy(scene["group_ids"]).unsqueeze(0).to(DEVICE),
    )


def ema_update(teacher,student,tau):
    with torch.no_grad():
        for pt,ps in zip(teacher.parameters(),student.parameters()):
            pt.data.mul_(tau).add_(ps.data,alpha=1-tau)


def rollout_partition(model,z0,source_mu,source_cov,gids,parts):
    z,mu,cov = z0,source_mu,source_cov
    elapsed = 0
    zs,geoms = [],[]
    for dlt in parts:
        h = torch.full((z.shape[0],),float(dlt),device=z.device)
        zn = model.transition(z,h)
        geom = model.decoder(z,zn,mu,cov,gids,h)
        elapsed += dlt
        z,mu,cov = zn,geom["mu"],geom["cov"]
        zs.append((elapsed,z))
        geoms.append((elapsed,geom))
    return zs,geoms


# ============================================================
# Validation and training
# ============================================================

@torch.no_grad()
def evaluate_target_accuracy(model,ds,cfg):
    model.eval()
    loader = DataLoader(ds,batch_size=cfg.eval_batch_size,shuffle=False,num_workers=0)
    direct_vals,rollout_vals = [],[]
    for batch in loader:
        batch = {k:(v.to(DEVICE) if torch.is_tensor(v) else v) for k,v in batch.items()}
        B = batch["mu"].shape[0]
        t,gids = batch["t"],batch["group_ids"]
        bidx = torch.arange(B,device=DEVICE)
        ctx_mu,ctx_cov = get_hist(batch,t,cfg)
        z0 = model.encoder(ctx_mu,ctx_cov,gids)
        src_mu,src_cov = batch["mu"][bidx,t],batch["cov"][bidx,t]

        for H in cfg.train_horizons:
            h = torch.full((B,),float(H),device=DEVICE)
            zd = model.transition(z0,h)
            gd = model.decoder(z0,zd,src_mu,src_cov,gids,h)
            gt_mu = batch["mu"][bidx,t+H]
            direct_vals.append(torch.linalg.norm(gd["mu"]-gt_mu,dim=-1).mean(-1).cpu())
            _,geoms = rollout_partition(model,z0,src_mu,src_cov,gids,PRIMARY_PARTITIONS[H])
            rollout_vals.append(torch.linalg.norm(geoms[-1][1]["mu"]-gt_mu,dim=-1).mean(-1).cpu())

    direct,rollout = torch.cat(direct_vals).mean().item(),torch.cat(rollout_vals).mean().item()
    return {"val_score":0.5*(direct+rollout),"val_direct_fde":direct,"val_rollout_fde":rollout}


def train_variant(variant,seed,cfg,train_ds,val_ds):
    seed_all(seed)
    model = JEPA(cfg).to(DEVICE)
    teacher = HierEncoder(cfg).to(DEVICE)
    teacher.load_state_dict(model.encoder.state_dict())
    for p in teacher.parameters():
        p.requires_grad_(False)
    teacher.eval()

    opt = torch.optim.AdamW(model.parameters(),lr=cfg.lr,weight_decay=cfg.weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=cfg.epochs,eta_min=cfg.min_lr)
    loader = DataLoader(train_ds,batch_size=cfg.batch_size,shuffle=True,num_workers=0,drop_last=True)
    use_path,use_comp = "Path" in variant,"Composition" in variant
    partition_rng = random.Random(100_000+seed)

    history,best_val,best_epoch = [],float("inf"),-1
    best_model_state,best_teacher_state = None,None
    global_step = 0

    for epoch in range(cfg.epochs):
        model.train()
        sums = {k:0.0 for k in ["total","end","geo","path","comp","rep"]}
        nstep = 0

        for batch in loader:
            batch = {k:(v.to(DEVICE) if torch.is_tensor(v) else v) for k,v in batch.items()}
            B = batch["mu"].shape[0]
            t,gids = batch["t"],batch["group_ids"]
            bidx = torch.arange(B,device=DEVICE)
            ctx_mu,ctx_cov = get_hist(batch,t,cfg)
            z0 = model.encoder(ctx_mu,ctx_cov,gids)

            Delta = cfg.train_horizons[global_step % len(cfg.train_horizons)]
            global_step += 1
            hD = torch.full((B,),float(Delta),device=DEVICE)
            tgt_end = t+Delta
            assert int(tgt_end.max()) < train_ds.heldout_idx

            tgt_mu_hist,tgt_cov_hist = get_hist(batch,tgt_end,cfg)
            with torch.no_grad():
                zt = teacher(tgt_mu_hist,tgt_cov_hist,gids)
            zd = model.transition(z0,hD)
            l_end = hier_distance(zd,zt,cfg)

            src_mu,src_cov = batch["mu"][bidx,t],batch["cov"][bidx,t]
            gt_mu,gt_cov = batch["mu"][bidx,tgt_end],batch["cov"][bidx,tgt_end]
            gd = model.decoder(z0,zd,src_mu,src_cov,gids,hD)
            l_geo = geometry_loss(gd["mu"],gd["cov"],gt_mu,gt_cov,cfg)
            l_rep = 0.5*(rep_regularizer(z0,cfg)+rep_regularizer(zd,cfg))

            l_path = torch.zeros((),device=DEVICE)
            l_comp = torch.zeros((),device=DEVICE)
            if use_path or use_comp:
                parts = partition_rng.choice(TRAIN_PARTITIONS[Delta])
                zs,_ = rollout_partition(model,z0,src_mu,src_cov,gids,parts)

                if use_path:
                    vals = []
                    for elapsed,zj in zs:
                        mh,ch = get_hist(batch,t+elapsed,cfg)
                        with torch.no_grad():
                            ztj = teacher(mh,ch,gids)
                        vals.append(hier_distance(zj,ztj,cfg))
                    l_path = torch.stack(vals).mean()

                if use_comp:
                    vals = []
                    for elapsed,zj in zs:
                        hj = torch.full((B,),float(elapsed),device=DEVICE)
                        zdir = model.transition(z0,hj)
                        vals.append(0.5*(hier_distance(zdir,zj.detach(),cfg)+hier_distance(zj,zdir.detach(),cfg)))
                    l_comp = torch.stack(vals).mean()

            total = l_end + cfg.lambda_geo*l_geo + cfg.lambda_path*l_path + cfg.lambda_comp*l_comp + cfg.lambda_rep*l_rep
            opt.zero_grad(set_to_none=True)
            total.backward()
            nn.utils.clip_grad_norm_(model.parameters(),cfg.grad_clip)
            opt.step()
            ema_update(teacher,model.encoder,cfg.ema_tau)

            for k,v in [("total",total),("end",l_end),("geo",l_geo),("path",l_path),("comp",l_comp),("rep",l_rep)]:
                sums[k] += float(v.detach())
            nstep += 1

        val = evaluate_target_accuracy(model,val_ds,cfg)
        row = {"epoch":epoch+1,"lr":opt.param_groups[0]["lr"],**{f"train_{k}":v/max(1,nstep) for k,v in sums.items()},**val}
        history.append(row)
        if val["val_score"] < best_val:
            best_val,best_epoch = val["val_score"],epoch+1
            best_model_state = {k:v.detach().cpu().clone() for k,v in model.state_dict().items()}
            best_teacher_state = {k:v.detach().cpu().clone() for k,v in teacher.state_dict().items()}
        print(f"[{variant} seed={seed}] epoch {epoch+1:02d}/{cfg.epochs} train={row['train_total']:.4f} val={val['val_score']:.4f} best={best_epoch}")
        scheduler.step()

    model.load_state_dict(best_model_state)
    teacher.load_state_dict(best_teacher_state)
    model.eval(); teacher.eval()
    hist = pd.DataFrame(history)
    return model,teacher,hist,{
        "best_epoch":best_epoch,
        "best_val_score":best_val,
        "best_val_direct_fde":float(hist.loc[hist.epoch==best_epoch,"val_direct_fde"].iloc[0]),
        "best_val_rollout_fde":float(hist.loc[hist.epoch==best_epoch,"val_rollout_fde"].iloc[0]),
    }


# ============================================================
# Held-out terminal-frame evaluation
# ============================================================

@torch.no_grad()
def predict_next(model,scene,cfg):
    target_idx = cfg.seq_len-1
    context_end = target_idx-1
    mh,ch,gids = scene_hist(scene,context_end,cfg)
    z0 = model.encoder(mh,ch,gids)
    h = torch.ones(1,device=DEVICE)
    z1 = model.transition(z0,h)
    src_mu = torch.from_numpy(scene["mu"][context_end]).unsqueeze(0).to(DEVICE)
    src_cov = torch.from_numpy(scene["cov"][context_end]).unsqueeze(0).to(DEVICE)
    geom = model.decoder(z0,z1,src_mu,src_cov,gids,h)
    return z0,z1,geom


@torch.no_grad()
def predict_h4(model,scene,cfg):
    target_idx = cfg.seq_len-1
    context_end = target_idx-4
    mh,ch,gids = scene_hist(scene,context_end,cfg)
    z0 = model.encoder(mh,ch,gids)
    src_mu = torch.from_numpy(scene["mu"][context_end]).unsqueeze(0).to(DEVICE)
    src_cov = torch.from_numpy(scene["cov"][context_end]).unsqueeze(0).to(DEVICE)
    h = torch.full((1,),4.0,device=DEVICE)
    zdir = model.transition(z0,h)
    gdir = model.decoder(z0,zdir,src_mu,src_cov,gids,h)
    zs,geoms = rollout_partition(model,z0,src_mu,src_cov,gids,[2,2])
    return zdir,gdir,zs[-1][1],geoms[-1][1]


@torch.no_grad()
def evaluate_baselines(test_scenes,cfg,lpips_net):
    rows = []
    T = cfg.seq_len-1
    for si,scene in enumerate(test_scenes):
        mu = torch.from_numpy(scene["mu"]).to(DEVICE)
        cov = torch.from_numpy(scene["cov"]).to(DEVICE)
        rgb = torch.from_numpy(scene["rgb"]).to(DEVICE)
        opacity = torch.from_numpy(scene["opacity"]).to(DEVICE)
        gt_mu,gt_cov = mu[T],cov[T]
        target = render_gaussians(gt_mu,gt_cov,rgb,opacity,cfg)
        current,previous = mu[T-1],mu[T-2]
        for method,pred_mu,pred_cov in [
            ("Last-frame",current,cov[T-1]),
            ("Constant velocity",current+(current-previous),cov[T-1]),
        ]:
            pred = render_gaussians(pred_mu,pred_cov,rgb,opacity,cfg)
            rows.append({
                "method":method,"scene_idx":si,
                "next_fde":float(torch.linalg.norm(pred_mu-gt_mu,dim=-1).mean().cpu()),
                "next_cov_err":float(covariance_error_per_sample(pred_cov[None],gt_cov[None])[0].cpu()),
                "next_mse":image_mse(pred,target),"next_psnr":image_psnr(pred,target),
                "next_ssim":image_ssim(pred,target),"next_lpips":lpips_score(lpips_net,pred,target),
            })
    return pd.DataFrame(rows)


@torch.no_grad()
def evaluate_model_heldout(model,variant,seed,test_scenes,cfg,lpips_net):
    rows,h4rows = [],[]
    T = cfg.seq_len-1
    for si,scene in enumerate(test_scenes):
        gt_mu = torch.from_numpy(scene["mu"][T]).to(DEVICE)
        gt_cov = torch.from_numpy(scene["cov"][T]).to(DEVICE)
        rgb = torch.from_numpy(scene["rgb"]).to(DEVICE)
        opacity = torch.from_numpy(scene["opacity"]).to(DEVICE)
        target = render_gaussians(gt_mu,gt_cov,rgb,opacity,cfg)

        _,_,geom = predict_next(model,scene,cfg)
        pmu,pcov = geom["mu"][0],geom["cov"][0]
        pred = render_gaussians(pmu,pcov,rgb,opacity,cfg)
        rows.append({
            "variant":variant,"seed":seed,"scene_idx":si,
            "next_fde":float(torch.linalg.norm(pmu-gt_mu,dim=-1).mean().cpu()),
            "next_cov_err":float(covariance_error_per_sample(pcov[None],gt_cov[None])[0].cpu()),
            "next_mse":image_mse(pred,target),"next_psnr":image_psnr(pred,target),
            "next_ssim":image_ssim(pred,target),"next_lpips":lpips_score(lpips_net,pred,target),
        })

        zdir,gdir,zcomp,gcomp = predict_h4(model,scene,cfg)
        idir = render_gaussians(gdir["mu"][0],gdir["cov"][0],rgb,opacity,cfg)
        icomp = render_gaussians(gcomp["mu"][0],gcomp["cov"][0],rgb,opacity,cfg)
        h4rows.append({
            "variant":variant,"seed":seed,"scene_idx":si,
            "h4_direct_fde":float(torch.linalg.norm(gdir["mu"][0]-gt_mu,dim=-1).mean().cpu()),
            "h4_rollout_fde":float(torch.linalg.norm(gcomp["mu"][0]-gt_mu,dim=-1).mean().cpu()),
            "h4_latent_comp_err":float(hier_distance_per_sample(zdir,zcomp,cfg)[0].cpu()),
            "h4_geo_comp_err":float(torch.linalg.norm(gdir["mu"][0]-gcomp["mu"][0],dim=-1).mean().cpu()),
            "h4_render_comp_mse":image_mse(idir,icomp),
            "h4_direct_psnr":image_psnr(idir,target),"h4_rollout_psnr":image_psnr(icomp,target),
            "h4_direct_ssim":image_ssim(idir,target),"h4_rollout_ssim":image_ssim(icomp,target),
        })
    return pd.DataFrame(rows),pd.DataFrame(h4rows)


# ============================================================
# Qualitative outputs
# ============================================================

def img_np(x):
    return x.detach().permute(1,2,0).clamp(0,1).cpu().numpy()


@torch.no_grad()
def save_highlight(model,scene,cfg):
    T = cfg.seq_len-1
    rgb = torch.from_numpy(scene["rgb"]).to(DEVICE)
    opacity = torch.from_numpy(scene["opacity"]).to(DEVICE)
    prev = render_scene_frame(scene,T-1,cfg)
    target = render_scene_frame(scene,T,cfg)

    cur_mu = torch.from_numpy(scene["mu"][T-1]).to(DEVICE)
    cur_cov = torch.from_numpy(scene["cov"][T-1]).to(DEVICE)
    prev_mu = torch.from_numpy(scene["mu"][T-2]).to(DEVICE)
    static = render_gaussians(cur_mu,cur_cov,rgb,opacity,cfg)
    cv = render_gaussians(cur_mu+(cur_mu-prev_mu),cur_cov,rgb,opacity,cfg)
    _,_,geom = predict_next(model,scene,cfg)
    pred = render_gaussians(geom["mu"][0],geom["cov"][0],rgb,opacity,cfg)

    fig,axes = plt.subplots(1,5,figsize=(15,3.1))
    for ax,im,title in zip(
        axes,[prev,static,cv,pred,target],
        [r"Last observed $I_{T-1}$","Last-frame","Constant velocity",r"4DGS-JEPA $\hat I_T$",r"Held-out $I_T$"]
    ):
        ax.imshow(img_np(im)); ax.set_title(title,fontsize=9); ax.axis("off")
    plt.tight_layout()
    plt.savefig(RESULT_DIR/"heldout_next_frame_highlight.png",dpi=220,bbox_inches="tight")
    plt.show()

    zdir,gdir,zcomp,gcomp = predict_h4(model,scene,cfg)
    start = render_scene_frame(scene,T-4,cfg)
    idir = render_gaussians(gdir["mu"][0],gdir["cov"][0],rgb,opacity,cfg)
    icomp = render_gaussians(gcomp["mu"][0],gcomp["cov"][0],rgb,opacity,cfg)
    fig,axes = plt.subplots(1,4,figsize=(12,3.1))
    for ax,im,title in zip(
        axes,[start,idir,icomp,target],
        [r"Context $I_{T-4}$",r"Direct $\Delta=4$",r"Composed $(2,2)$",r"Held-out $I_T$"]
    ):
        ax.imshow(img_np(im)); ax.set_title(title,fontsize=9); ax.axis("off")
    plt.tight_layout()
    plt.savefig(RESULT_DIR/"heldout_h4_composition_highlight.png",dpi=220,bbox_inches="tight")
    plt.show()


def save_ground_truth_gif(scene,cfg):
    try:
        import imageio.v2 as imageio
    except Exception as exc:
        print("imageio unavailable; GIF skipped:",exc)
        return
    frames = [(255*img_np(render_scene_frame(scene,t,cfg))).clip(0,255).astype(np.uint8) for t in range(cfg.seq_len)]
    imageio.mimsave(RESULT_DIR/"highlight_ground_truth_video.gif",frames,duration=0.12,loop=0)


# ============================================================
# Protocol checks
# ============================================================

def run_design_checks(train_seeds,val_seeds,test_seeds,train_ds,val_ds,test_scenes,cfg):
    assert set(train_seeds).isdisjoint(val_seeds)
    assert set(train_seeds).isdisjoint(test_seeds)
    assert set(val_seeds).isdisjoint(test_seeds)
    heldout = cfg.seq_len-1
    for ds in (train_ds,val_ds):
        for _,t in ds.samples:
            for H in cfg.train_horizons:
                assert t+H < heldout

    # Held-out-frame perturbation cannot change the causal context features.
    scene = test_scenes[0]
    end = heldout-1
    mh,ch,gids = scene_hist(scene,end,cfg)
    g1,m1 = causal_gaussian_features(mh,ch,gids,cfg)
    scene2 = copy.deepcopy(scene)
    scene2["mu"][heldout] += 999
    scene2["cov"][heldout] *= 50
    mh2,ch2,_ = scene_hist(scene2,end,cfg)
    g2,m2 = causal_gaussian_features(mh2,ch2,gids,cfg)
    assert torch.equal(g1,g2) and torch.equal(m1,m2)

    print("Protocol checks passed:")
    print("  ✓ disjoint train/validation/test scene seeds")
    print("  ✓ final frame excluded from every training/validation target")
    print("  ✓ student context cannot depend on held-out terminal frame")
    print("  ✓ rendering is evaluation-only; no RGB loss is optimized")
    print("  ✓ training horizons {1,2,4} retain non-trivial temporal composition")


# ============================================================
# Run
# ============================================================

train_seeds,val_seeds,test_seeds = build_splits(CFG)
train_ds = GaussianVideoTrainDataset(train_seeds,CFG)
val_ds = GaussianVideoTrainDataset(val_seeds,CFG)
test_scenes = [generate_scene(s,CFG) for s in test_seeds]
print("Train samples:",len(train_ds),"Val samples:",len(val_ds),"Held-out test frames:",len(test_scenes))

run_design_checks(train_seeds,val_seeds,test_seeds,train_ds,val_ds,test_scenes,CFG)
lpips_net = maybe_build_lpips(CFG)
save_ground_truth_gif(test_scenes[CFG.highlight_scene_index],CFG)

baseline_df = evaluate_baselines(test_scenes,CFG,lpips_net)
baseline_df.to_csv(RESULT_DIR/"heldout_baseline_per_scene.csv",index=False)
baseline_summary = baseline_df.groupby("method")[["next_fde","next_cov_err","next_mse","next_psnr","next_ssim","next_lpips"]].agg(["mean","std"])
baseline_summary.to_csv(RESULT_DIR/"heldout_baseline_summary.csv")
print("\nBaselines"); display(baseline_summary)

VARIANTS = ["Endpoint","Endpoint + Path","Endpoint + Composition","Endpoint + Path + Composition"]
model_rows,h4_rows,run_infos = [],[],[]

for variant in VARIANTS:
    for seed in CFG.seeds:
        model,teacher,hist,best_info = train_variant(variant,seed,CFG,train_ds,val_ds)
        stem = variant.replace(" ","_").replace("+","plus")
        hist.to_csv(RESULT_DIR/f"history_{stem}_seed{seed}.csv",index=False)
        mdf,hdf = evaluate_model_heldout(model,variant,seed,test_scenes,CFG,lpips_net)
        model_rows.append(mdf); h4_rows.append(hdf)
        run_infos.append({"variant":variant,"seed":seed,**best_info})

        torch.save(
            {"model":{k:v.detach().cpu() for k,v in model.state_dict().items()},
             "teacher":{k:v.detach().cpu() for k,v in teacher.state_dict().items()},
             "config":asdict(CFG),**best_info},
            RESULT_DIR/f"best_model_{stem}_seed{seed}.pt"
        )

        if variant==CFG.highlight_variant and seed==CFG.highlight_seed:
            save_highlight(model,test_scenes[CFG.highlight_scene_index],CFG)

        del model,teacher
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

model_df = pd.concat(model_rows,ignore_index=True)
h4_df = pd.concat(h4_rows,ignore_index=True)
run_info_df = pd.DataFrame(run_infos)
model_df.to_csv(RESULT_DIR/"heldout_model_per_scene.csv",index=False)
h4_df.to_csv(RESULT_DIR/"heldout_h4_composition_per_scene.csv",index=False)
run_info_df.to_csv(RESULT_DIR/"checkpoint_summary.csv",index=False)

MAIN = ["next_fde","next_cov_err","next_mse","next_psnr","next_ssim","next_lpips"]
per_seed = model_df.groupby(["variant","seed"])[MAIN].mean().reset_index()
model_summary = per_seed.groupby("variant")[MAIN].agg(["mean","std"])
model_summary.to_csv(RESULT_DIR/"heldout_model_summary_mean_std.csv")

H4 = ["h4_direct_fde","h4_rollout_fde","h4_latent_comp_err","h4_geo_comp_err",
      "h4_render_comp_mse","h4_direct_psnr","h4_rollout_psnr","h4_direct_ssim","h4_rollout_ssim"]
h4_per_seed = h4_df.groupby(["variant","seed"])[H4].mean().reset_index()
h4_summary = h4_per_seed.groupby("variant")[H4].agg(["mean","std"])
h4_summary.to_csv(RESULT_DIR/"heldout_h4_summary_mean_std.csv")

print("\nHeld-out next-frame learned models"); display(model_summary)
print("\nHeld-out Delta=4 direct/composed diagnostic"); display(h4_summary)
print("\nCheckpoint summary"); display(run_info_df)

# Compact paper tables
baseline_df.groupby("method")[["next_fde","next_psnr","next_ssim","next_lpips"]].agg(["mean","std"]).to_csv(
    RESULT_DIR/"paper_table_baselines.csv"
)
per_seed.groupby("variant")[["next_fde","next_psnr","next_ssim","next_lpips"]].agg(["mean","std"]).to_csv(
    RESULT_DIR/"paper_table_models.csv"
)

# Simple plots
for metric,ylabel in [
    ("next_psnr","Held-out next-frame PSNR (dB)"),
    ("next_ssim","Held-out next-frame SSIM"),
    ("next_fde","Held-out next-frame Gaussian FDE"),
]:
    stats = per_seed.groupby("variant")[metric].agg(["mean","std"])
    plt.figure(figsize=(8,4.5))
    xs = np.arange(len(stats))
    plt.bar(xs,stats["mean"].values,yerr=stats["std"].values,capsize=4)
    plt.xticks(xs,stats.index,rotation=20,ha="right")
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.savefig(RESULT_DIR/f"{metric}.png",dpi=180)
    plt.show()

stats = h4_per_seed.groupby("variant")["h4_render_comp_mse"].agg(["mean","std"])
plt.figure(figsize=(8,4.5))
xs = np.arange(len(stats))
plt.bar(xs,stats["mean"].values,yerr=stats["std"].values,capsize=4)
plt.xticks(xs,stats.index,rotation=20,ha="right")
plt.ylabel(r"Held-out $\Delta=4$ direct--composed render MSE")
plt.tight_layout()
plt.savefig(RESULT_DIR/"h4_render_comp_mse.png",dpi=180)
plt.show()

print("\nExperiment 4 completed.")
print("Results:",RESULT_DIR)
print("Main evaluation: final held-out frame, Delta=1.")
print("Secondary diagnostic: same held-out frame from Delta=4, direct vs composed (2,2).")
print("No RGB/rendering loss was used for optimization.")
```
